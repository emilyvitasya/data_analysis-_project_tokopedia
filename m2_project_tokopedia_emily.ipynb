### 2.2 Kerangka Identifikasi Missing Values

Sebelum menangani missing values pada kolom manapun, identifikasi dulu **jenis missing value-nya**. Jenis ini menentukan metode penanganan yang valid.

| Jenis | Definisi Singkat | Implikasi Penanganan | Contoh di Dataset Ini |
|---|---|---|---|
| **MCAR** *(Missing Completely At Random)* | Nilai kosong tidak berkaitan dengan kolom lain. Pola missing benar-benar acak. | Relatif aman di-impute (mean/median/modus) atau di-drop tanpa bias signifikan. | `gender` kosong secara acak tanpa pola tertentu. |
| **MAR** *(Missing At Random)* | Nilai kosong berkaitan dengan kolom **lain**, bukan dengan nilai kolom itu sendiri. | Imputation berbasis kolom lain lebih tepat. Drop bisa menyebabkan bias. | `courier_name` kosong mungkin berkaitan dengan tanggal atau metode pembayaran tertentu. |
| **MNAR** *(Missing Not At Random)* | Nilai kosong berkaitan langsung dengan nilai yang seharusnya ada. Ada alasan sistematis. | Imputation apapun berisiko misleading. Perlu keputusan bisnis eksplisit. | `courier_name` kosong karena order dibatalkan sebelum pickup — nilai kosong itu sendiri adalah informasi. |

> 💡 **Cara Menggunakan Kerangka Ini:**
> Untuk setiap kolom bermasalah, tanyakan:
> 1. Apakah pola missing-nya acak, atau ada pola tertentu?
> 2. Apakah nilai kosong berkaitan dengan kolom lain?
> 3. Apakah nilai kosong itu sendiri mengandung informasi bisnis?
>
> Justifikasi reasoning kamu lebih penting dari labelnya.

```python
# TODO: Standarisasi payment_method
# - apabila provider paymentnya sama, namun penulisannya berbeda maka bisa disatukan, namun apabila provider paymentnya
# berbeda, harus dipisahkan dengan standar penulisan masing-masing. Alasan mandiri VA dan BCA VA harus dipisah adalah
# untuk alasan permintaan promosi/cashback apabila penggunanya banyak
```

```python
table_3=table_check_1.copy()
table_3.replace(to_replace=r"[Gg][Oo][\s\-]?[Pp][Aa][Yy]", value="GoPay", regex=True, inplace=True)
table_3.replace(to_replace=r"[Bb][Cc][Aa][\s]([Vv][Aa]|[Vv]irtual [Aa]ccount)", value="BCA VA", regex=True, inplace=True)
table_3.replace(to_replace=r"[Oo][Vv][Oo]", value="OVO", regex=True, inplace=True)
table_3["payment_method"].unique()
```

```python
table_3 = table_check_1.copy()
table_3["payment_method"] = table_3["payment_method"].replace(
    to_replace=r"[Gg][Oo][\s\-_]?[Pp][Aa][Yy]", value="GoPay", regex=True
)
table_3["payment_method"] = table_3["payment_method"].replace(
    to_replace=r"[Bb][Cc][Aa][\s\-_]?([Vv][Aa]|[Vv]irtual[\s]?[Aa]ccount)", value="BCA VA", regex=True
)
table_3["payment_method"] = table_3["payment_method"].replace(
    to_replace=r"[Mm][Aa][Nn][Dd][Ii][Rr][Ii][\s\-_]?([Vv][Aa]|[Vv]irtual[\s]?[Aa]ccount)", value="Mandiri VA", regex=True
)
table_3["payment_method"] = table_3["payment_method"].replace(
    to_replace=r"[Oo][Vv][Oo]", value="OVO", regex=True
)
print(table_3["payment_method"].unique())
```
