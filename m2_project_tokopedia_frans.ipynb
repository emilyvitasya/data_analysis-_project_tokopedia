## 0. Import & Load Data

```python
import pandas as pd
import numpy as np

pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 50)
pd.set_option('display.float_format', '{:.2f}'.format)
```

```python
# Load semua dataset
# Sesuaikan path dengan lokasi file kamu
df_users_raw = pd.read_csv('tokopedia_users.csv')
df_trans_raw = pd.read_csv('tokopedia_transactions.csv')
df_deliv_raw = pd.read_csv('tokopedia_deliveries.csv')

# Buat copy untuk dikerjakan
df_users = df_users_raw.copy()
df_trans = df_trans_raw.copy()
df_deliv = df_deliv_raw.copy()

print(f'users    : {df_users.shape}')
print(f'transactions : {df_trans.shape}')
print(f'deliveries   : {df_deliv.shape}')
```
